```python
%python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(u"I will google about facebook")
```

```python
doc.text
```

```python
for token in doc:
    print(token, token.pos_, token.tag_, spacy.explain(token.tag_))
```
