```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.count = 0  # number of words passing through
        self.is_end = False
```

```python
class Trie:
    def __init__(self):
        self.root = TrieNode()

    # Method to insert a key into the Trie
    def insert(self, word):
        node = self.root
        node.count += 1
        for ch in word:
            if ch not in node.children:
                node.children[ch] = TrieNode()
            node = node.children[ch]
            node.count += 1
        node.is_end = True

    # Method to search a key in the Trie
    def search(self, key):
        curr = self.root
        for c in key:
            index = ord(c) - ord('a')
            if curr.children[index] is None:
                return False
            curr = curr.children[index]
        return curr.isLeaf

    # Method to check if a prefix exists in the Trie
    def isPrefix(self, prefix):
        curr = self.root
        for c in prefix:
            index = ord(c) - ord('a')
            if curr.children[index] is None:
                return False
            curr = curr.children[index]
        return True

    def stem_and_suffix(self, word):
        """
        Traverse along the word, pick point where branching factor is max as suffix start.
        """
        node = self.root
        max_branch_node_idx = 0
        max_branch_count = 3
        
        for i, ch in enumerate(word):
            if ch not in node.children:
                return word, ""  # not found fully
            node = node.children[ch]
            branching = len(node.children)
            if branching > max_branch_count:
                max_branch_count = branching
                max_branch_node_idx = i + 1  # stem ends here
        
        stem = word[:max_branch_node_idx]
        suffix = word[max_branch_node_idx:]
        return stem, suffix
```

```python
def load_words(filename):
    with open(filename, "r", encoding="utf-8") as f:
        return [line.strip().lower() for line in f if line.strip()]
```

```python
filename = "brown_nouns.txt"
try:
    words = load_words(filename)
except FileNotFoundError:
    words = ["goes", "kite", "kites", "dogs", "foxes", "churches", "tries", "buses"]
```

```python
prefix_trie = Trie()
for w in words:
    prefix_trie.insert(w)
```

```python
suffix_trie = Trie()
for w in words:
    suffix_trie.insert(w[::-1])
```

```python
# Compare stem+suffix results
results = []
for w in words:
    pre_stem, pre_suff = prefix_trie.stem_and_suffix(w)
    suf_stem_rev, suf_suff_rev = suffix_trie.stem_and_suffix(w[::-1])
    suf_suff = suf_stem_rev[::-1]  # reversed because suffix trie splits reversed word
    suf_stem = w[:len(w) - len(suf_suff)]
    
    results.append(f"{w} (prefix trie)={pre_stem}+{pre_suff} | (suffix trie)={suf_stem}+{suf_suff}")
```

```python
for r in results:
    print(r)
```
